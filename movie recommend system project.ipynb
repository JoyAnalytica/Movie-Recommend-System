```python
import pandas as pd
import numpy as np
import warnings
import nltk
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv('tmdb_5000_movies.csv')
cre=pd.read_csv('tmdb_5000_credits.csv')
```

```python
df.head(2)
```

```python
cre.head(2)
```

```python
# merge two tables

movie=df.merge(cre,on='title')
```

```python
movie.head(1)
```

```python
# budget
# homepage
# id
# original_language
# original_title
# popularity
# production_comapny
# production_countries
# release-date(not sure)
```

```python
movie = movie[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
movie.head()
```

```python
import ast
```

```python
def recommend(text):
    l=[]
    for i in ast.literal_eval(text):
        l.append(i['name'])
    return l
```

```python
movie['genres']=movie['genres'].apply(recommend)
```

```python
movie['keywords']=movie['keywords'].apply(recommend)
```

```python
movie['cast']=movie['cast'].apply(recommend)
```

```python
movie['cast']=movie['cast'].apply(lambda x:x[0:3])
```

```python
def crew_fetch(text):
    L=[]
    for i in ast.literal_eval(text):
        if i["job"]=="Director":
            L.append(i['name'])
    return L
```

```python
movie['crew']=movie['crew'].apply(crew_fetch)
```

```python
def space_replace(text):
    L1=[]
    for i in text:
        L1.append(i.replace(" ",""))
    return L1
        
```

```python
movie['genres']=movie['genres'].apply(space_replace)
movie['keywords']=movie['keywords'].apply(space_replace)
movie['cast']=movie['cast'].apply(space_replace)
movie['crew']=movie['crew'].apply(space_replace)
```

```python
movie.dropna(inplace=True)
```

```python
movie['overview']=movie['overview'].apply(lambda x:x.split())
```

```python
movie['tags'] = movie['overview'] + movie['genres'] + movie['keywords'] + movie['cast'] + movie['crew']
```

```python
movie['tags']=movie['tags'].apply(lambda x:' '.join(x))
```

```python
new = movie.drop(columns=['overview','genres','keywords','cast','crew'])
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
```

```python
new['tags']=new['tags'].apply(lambda x:x.split())
```

```python
new['tags']=new['tags'].apply(lambda x:[ps.stem(word) for word in x])
```

```python
new['tags']=new['tags'].apply(lambda x:' '.join(x))
```

```python
new['tags']=new['tags'].apply(lambda x:x.lower())
```

# Model building

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
```

```python
cv=CountVectorizer(max_features=5000)
tfidf=TfidfVectorizer(max_features=5000)
```

```python
x=cv.fit_transform(new['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(x)
```

```python
new[new['title']=='Avatar'].index[0]
```

```python
def recommends(movie):
    index = new[new['title'] == movie].index[0]

    distance = sorted(
        list(enumerate(similarity[index])),
        reverse=True,
        key=lambda x: x[1]
    )

    for i in distance[1:6]:
        print(new.iloc[i[0]].title)
```

```python
recommends('Battleship')
```
